```python
import random

all_acc = []

def open_account(cnic, deposit=0):
  acc_title = input("Enter account title: ")
  acc_num = random.randint(10000,99999)
  acc_pin = random.randint(1000,9999)
  account = {
    "Account_Title":acc_title,
    "Account_Num":acc_num,
    "Account_Pin":acc_pin,
    "Balance":deposit
  }
  all_acc.append(account)
  return f"Account opened... {account}"


open_account(123456789,10000)
```

```python
def check_balance(acc_num):
  for acc in all_acc:
    if acc['Account_Num'] == acc_num:
      pin = int(input("Enter your account pin: "))
      if pin == acc['Account_Pin']:
        print(f"Your balance is: {acc['Balance']}")
      else:
        print("You entered the wrong pin...")
    else:
      print("You entered the wrong account number...")

check_balance(86739)
```

```python
def cash_withdrawl(acc_num):
  for acc in all_acc:
    if acc['Account_Num'] == acc_num:
      pin = int(input("Enter your account pin: "))
      if pin == acc['Account_Pin']:
        withdraw = int(input("How much cash do you wanna withdraw: "))
        acc['Balance'] -= withdraw
        print(f"Now your balance is: {acc['Balance']}")
      else:
        print("You entered the wrong pin...")
    else:
      print("You entered the wrong account number...")

cash_withdrawl(86739)
```

```python
def cash_deposit(acc_num):
  for acc in all_acc:
    if acc_num == acc['Account_Num']:
      print(f"Your current balance is: {acc['Balance']}")
      user = int(input("How much cash do you wanna deposit: "))
      acc['Balance'] += user
      print(f"Your new balance is: {acc['Balance']}")
    else:
      print("Sorry, we dont have an account with this account number")


cash_deposit(86739)
```
