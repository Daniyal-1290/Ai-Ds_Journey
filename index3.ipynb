## EXCEPTION HANDLING

```python
try:
    num1 = int(input("Enter the 1st number: "))
    num2 = int(input("Enter the 2nd number: "))
    print(num1/num2)
except ZeroDivisionError:
    print("You cannot divide a number by zero")
except ValueError:
    print("Please enter a number value...")
```

```python
try:
    num1 = int(input("Enter the 1st number: "))
    num2 = int(input("Enter the 2nd number: "))
    print(num1/num2)
except Exception  as e:
    print(f"Error handled {e}")
```

## FILE READ WRITE I/O

```python
try:
    file = open('index.ipynb', 'r')
    for line in file:
        print(line, end="")

    file.close()
except FileNotFoundError:
    print("File not found")
```

```python
with open('index1.ipynb', 'r') as file:
    for line in file:
        print(line, end="")
```

```python
with open('index.txt', 'w') as writingfile:
    writingfile.write("Hello broo\n")
```

```python
with open('index.txt', 'a') as writingfile:
    writingfile.write("How are you...")
```

```python
with open('students.txt', 'w') as students:
    while True:
        name = input("Enter students name or q to quit: ")
        if name == "q":
            break
        else:
            students.write(f"{name} \n")
```
