```python
import matplotlib.pyplot as plt
```

```python
x = [2023, 2024, 2025, 2026]
y1 = [15, 25, 30, 20]
y2 = [17, 23, 38, 5]

plotstyle = dict(marker='.',markersize='30', markerfacecolor='#1cd3fc', markeredgecolor='#1cd3fc', linestyle='dashed', linewidth=4)

plt.title("Class size", fontsize=25, family='Arial', fontweight='bold', color='#2d4cfc')
plt.xlabel("Year", fontsize=20, family='Arial', fontweight='bold', color='#2dbefc')
plt.ylabel("Students", fontsize=20, family='Arial', fontweight='bold', color='#2dbefc')
plt.xticks(x, rotation=10)
plt.tick_params(axis='both', colors='#2d4cfc')
plt.grid(axis='y', linewidth=2, color='lightgray', linestyle='dashed')

plt.plot(x,y1, color='#fc1c1c', **plotstyle, label='A')
plt.plot(x, y2, color='#1cfc45', **plotstyle, label='B')
plt.legend()
plt.show()
```

## BAR CHARTS

```python
categories = ["Fruits", "Vegetables", "Dairy", "Sweets",]
values = [20, 25, 10, 40]

plt.title("SHOP", fontsize=20, fontweight='bold', family='Arial')
plt.grid(axis='y')
plt.xlabel("Food", fontsize=20)
plt.ylabel("Quantity", fontsize=20)
 
plt.bar(categories, values, color='skyblue') #this is  for vertical barchart
# plt.barh(categories, values, color='skyblue')  this is for horizontal barchart
plt.show()
```

## PIE CHART

```python
categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values = [300, 250, 275, 225]
colors = ["red", "yellow", "blue", "green"]

plt.title("Daniyal college")

plt.pie(values, labels=categories, autopct="%1.1f%%", colors=colors, explode=[0,0,0,0.1], shadow=True, startangle=90)
plt.show()
```

## SCATTER PLOTS

```python
""" Shows the relationship between two variables
    Helps to identify a correlation (+, -, None)
    Example: Studyhours vs Testscores
"""


studyhours = [0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]
testscores = [55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]

plt.xlabel("Hours studied")
plt.ylabel("Test Score")
plt.title("Result")

plt.scatter(studyhours, testscores, color='blue', alpha=0.5, s=100, label='A') #alpha is used for transperancy #s is used for size #if label is used then also use legend func
plt.legend()
plt.show()
```

## HISTOGRAMS

```python
marks = [45, 56, 67, 78, 89, 90, 67, 76, 55, 45,
         88, 92, 73, 65, 80, 85, 70, 60, 95, 50]

plt.hist(marks, bins=5, edgecolor='black')

plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Distribution of Student Marks')

plt.show()
```
