## DATAFRAME

```python
data = {
    "Name":["Micheal", "Tim", "Harry", "Hassan"],
    "Class":["xi", "xi", "xi", "xi"],
    "Age":[16, 16, 17, 16]
}

df = pd.DataFrame(data)
df["Percentage"] = [76, 74, 70, 75]
newrow = pd.DataFrame([{"Name":"Hadi", "Class":"xi", "Age":17, "Percentage":80}], index=[4])
df = pd.concat([df, newrow])

df
```
