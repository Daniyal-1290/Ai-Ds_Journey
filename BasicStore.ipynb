```python
store = {
    'Earbuds':{'price':5000, 'quantity':2000},
    'Laptop':{'price':100000, 'quantity':1000},
    'Phone':{'price':80000, 'quantity':1500},
    'Speaker':{'price':15000, 'quantity':2500},
    'Monitor':{'price':30000, 'quantity':1000}
   }

user = input("What do you wanna buy: ").title()

for item in store:
  if item == user:
    quantity = int(input(f"How many {user} do you want: "))
    if quantity <= store[user]['quantity']:
      bill = quantity * store[user]['price']
      store[item]['quantity'] -= quantity
      print(f"Your bill is {bill}")
    else:
      print(f"We don't have {quantity} pieces")

    break

else:
  print(f"Sorry!, we dont have {user}")
```
