```python
cnic = input("Do you have cnic, yes or no: ")
if cnic == "yes":
  print("OK")
  election = input("Is today election day: ")
  if election == "yes":
    print("Ok")
    applied = input("Have you applied for vote registration: ")
    if applied == "yes":
      print("OK")
      print("You are eligible to vote...")
    else:
      print("Sorry you are not eligible to vote cuz you are not registered...")
  else:
    print("Its not election day today...")
else:
  print("Sorry you cannot vote cuz you dont have a cnic...")
```
