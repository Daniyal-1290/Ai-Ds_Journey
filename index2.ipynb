```python
def num(nums=[]):
  sum = 0
  for i in nums:
    sum+=i
  print(sum)

num([12,2,1,2,4])

```

```python
def oddnumsum(lst=[]):
  sum = 0
  for i in lst:
    if i % 2 != 0:
      sum += i
  print(sum)


oddnumsum([2,3,4,5,6,7,8,8,9,9,12,11])
```

```python
def num_exp(*n):
  for i in n:
    print(i**2, end=" ")

num_exp(1,2,3,4,5)
```

```python
def add_nums(a, *b):
  total = a
  for i in b:
    total+=i
  print(total)

add_nums(1,2,3,4)
```
