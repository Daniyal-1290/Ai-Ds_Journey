```python
class Student:
    def __init__(self,name,grade,age,gender):
        self.name = name
        self.grade = grade
        self.age = age
        self.gender = gender

    def profile(self):
        print(f"Name: {self.name}, Grade: {self.grade}, Age: {self.age}, Gender: {self.gender}")
        
```

```python
s1.profile()
s2.profile()
```

### ENCAPSULATION
- by default in classes all parameters(variables) and methods are public.
- to make a parameter protected, we must add single underscore before the variable name.
- to make a parameter private, we must add double underscores before the variable name. 

```python
class Laptop:
    def __init__(self, brand, ram, rom, processor):
        self.__brand = brand
        self._ram = ram
        self._rom = rom
        self._processor = processor

    def laptop_specs(self):
        print(f"{self.__brand}, {self._ram}, {self._rom}, {self._processor}")

    def laptop_price(self):
        if self._ram <= 8:
            print(100000)
        else:
            print(150000)
```

```python
laptop1 = Laptop('hp', 8, 256, "i5 10gen")
```

```python
laptop1.laptop_price()
```

```python
laptop1._brand = 'dell'
```

```python
laptop1.laptop_specs()
```

### INHERITANCE

```python
class Car:
    def __init__(self, color, speed, odometer):
        self._color = color
        self._speed = speed
        self._odometer = odometer

    def info(self):
        print(f" Color: {self._color}\n Speed: {self._speed}\n Odometer: {self._odometer}")
```

```python
car1 = Car("Black", 220, 20)
```

```python
car1.info()
```

```python
class ElectricCar(Car):
    def __init__(self, color, speed, odometer, battery, charger):
        super().__init__(color, speed, odometer)
        self._battery = battery
        self._charger = charger

    def info(self):
        print(f"""
        Color: {self._color}
        Speed: {self._speed}
        Odometer: {self._odometer}
        Battery: {self._battery}
        Charger: {self._charger}
        """)
```

```python
ecar1 = ElectricCar("Black", 230, 50, "100000 MAH", "Yes")
```

```python
ecar1.info()
```

### POLYMORPHISM

```python
class cartwo(Car):
    def __init__(self, color, speed, odometer):
        super().__init__(color, speed, odometer)

    def info(self):
        super().info()
```

```python
s1 = cartwo("Black", 220, 50)
```

```python
s1.info()
```

### ABSTRACTION

```python
from abc import ABC, abstractmethod
class Car(ABC):
    @abstractmethod
    def mileage(self):
        pass

class Bmw(Car):
    def mileage(self):
        print("Mileage is 15km per litre")

c1 = Bmw()
c1.mileage()
```
